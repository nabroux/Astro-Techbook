### 爬樓梯問題 
其實就是斐波那契

•如果最後一步走 1 階，那你前面一定已經在 n-1 階。
→ 所有走到 n-1 的方法都能「加一步」到 n。

•如果最後一步走 2 階，那你前面一定在 n-2 階。
→ 所有走到 n-2 的方法都能「加兩步」到 n。

```python
"""
有一座共有 n 階 的樓梯，
你一次可以走 1 階 或 2 階。
問：你有多少種不同走法能到達最上面？
"""

def climb_stairs(n):
    if n <= 2:
        return n

    dp = [0] * (n + 1)
    dp[1] = 1  # 走到第1階：1種方法
    dp[2] = 2  # 走到第2階：2種方法

    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]  # 狀態轉移

    return dp[n]

# 時間 O(n)
# 空間 O(n)
```

如果變成可以走1,2,3步會怎樣？


	•	你可能從 n-1 走 1 階上來，
	•	或從 n-2 走 2 階上來，
	•	或從 n-3 走 3 階上來。

所以就是    f(n) = f(n-1) + f(n-2) + f(n-3)

```python
def climb_123(n):
    # base cases
    if n <= 0:
        return 0
    if n == 1:
        return 1
    if n == 2:
        return 2
    if n == 3:
        return 4

    # dp[i] = 走到第 i 階的方法數
    dp = [0] * (n + 1)
    dp[1], dp[2], dp[3] = 1, 2, 4

    for i in range(4, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2] + dp[i - 3]

    return dp[n]
```

```python
# 空間優化版本
# 因為要求這題的解，只跟前三階有關，不需要整張表
# 優化後空間複雜度為 O(1)

def climb_123_fast(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    if n == 2:
        return 2
    if n == 3:
        return 4

    a, b, c = 1, 2, 4  # 對應 f(n-3), f(n-2), f(n-1)
    for _ in range(4, n + 1):
        a, b, c = b, c, a + b + c
    return c
```

### LIS
（Longest Increasing Subsequence，最長遞增子序列）

是 Dynamic Programming 中最經典、最常考、也最容易「一開始覺得難」的題之一。

我們來一步一步拆解它的邏輯、動態規劃思維，

並用 Python 代碼實現兩種版本（O(n²) 與 O(n log n)）。




```python
def length_of_LIS(nums):
    n = len(nums)
    if n == 0:
        return 0

    dp = [1] * n  # 每個數字至少能單獨成為長度1的序列

    for i in range(1, n):
        for j in range(0, i):
            if nums[j] < nums[i]:  # 能接在 j 後面
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)


nums = [10, 9, 2, 5, 3, 7, 101, 18]
print(length_of_LIS(nums))  # 4
```

### 「面額湊金額問題（Coin Change Problem）
給定幾種不同面額的硬幣 coins，以及一個總金額 amount。
求出湊出該金額所需的「最少硬幣數」。
若無法湊出，回傳 -1。



```python
def coin_change(coins, amount):
    # dp[i] = 湊出金額 i 所需的最少硬幣數
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for i in range(1, amount + 1):
        for c in coins:
            if i - c >= 0:
                dp[i] = min(dp[i], dp[i - c] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1

# 時間複雜度：O(n * m)，n 是金額，m 是硬幣種類數
# 空間複雜度：O(n)

coins = [1, 2, 5]
amount = 11
print(coin_change(coins, amount))  # 3


```
